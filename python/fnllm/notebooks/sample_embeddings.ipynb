Required imports:

```python
# Copyright (c) 2025 Microsoft Corporation.

import logging
import os

from dotenv import load_dotenv
from fnllm.events.logger import LLMEventsLogger
from fnllm.openai import (
    AzureOpenAIConfig,
    create_openai_embeddings_llm,
    create_openai_embeddings_rest_llm,
)

load_dotenv()
```

Example logger to showcase event logging:

```python
def _create_logger():
    logger = logging.getLogger("CHAT")
    logger.setLevel(logging.DEBUG)

    handler = logging.StreamHandler()
    handler.setLevel(logging.DEBUG)
    handler.setFormatter(
        logging.Formatter("[%(levelname)s %(asctime)s %(name)s] - %(message)s")
    )
    logger.handlers = [handler]

    return logger
```

Creating embeddings client:

```python
config = AzureOpenAIConfig(
    endpoint=os.environ["OPENAI_API_ENDPOINT"],
    api_version=os.environ["OPENAI_API_VERSION"],
    model=os.environ["OPENAI_API_MODEL"],
)

embeddings_client = create_openai_embeddings_llm(
    config,
    events=LLMEventsLogger(),
)
```

```python
res = await embeddings_client("Embed this for me")
```

```python
res.output.embeddings
```

```python
import httpx
TRANSPORT = httpx.AsyncHTTPTransport(http2=False, retries=2)
from dotenv import load_dotenv
load_dotenv()

http_client = httpx.AsyncClient(
            transport=TRANSPORT,
            timeout=httpx.Timeout(
                connect=15,  # keep connection timeout bounded
                read=120,
                write=120,
                pool=60.0,
            ),
            limits=httpx.Limits(
                max_connections=100,
                max_keepalive_connections=20,
                keepalive_expiry=15.0,  # slightly longer to reduce reconnect churn in containers
            ),
        )
url = "https://data-ai-dev.microsoft.com/metisaca/v1/embeddings"
body = {
    "input": ["Embed this for me"]
}
API_KEY = os.environ["METIS_API_KEY"]
headers = {
    "Content-Type": "application/json",
    "Authorization": f"Bearer {API_KEY}"
}
resp = await http_client.post(url, json=body, headers=headers)
```
